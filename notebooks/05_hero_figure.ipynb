# Milestone 5: The Hero Figure

This notebook plots the test error of every model we built, so one picture shows how far each one is off.

```python
import matplotlib.pyplot as plt

n_test = 9968  # len(test), from notebooks 02 and 03
results = {  # test RMSE, from notebooks 02 and 03
    "always 3": 1.2516,
    "overall average": 1.1349,
    "user average": 1.0426,
    "movie average": 1.0284,
    "user + movie effects": 0.95449,
    "low rank (k = 15)": 0.9250,
    "iterative SVD (k = 5)": 0.9206,
}
colors = ["tab:green" if "k =" in name else "gray" for name in results]

plt.figure(figsize=(8, 4))
bars = plt.barh(list(results), list(results.values()), color=colors)
plt.bar_label(bars, fmt="%.3f", padding=3)
plt.gca().invert_yaxis()
plt.xlabel("test RMSE (stars, lower is better)")
plt.title(f"Predicting {n_test:,} hidden MovieLens ratings")
plt.xlim(0, 1.6)
plt.tight_layout()
plt.savefig("hero.png", dpi=150)
plt.show()
```

This code draws one horizontal bar per model, gray for the simple baselines and green for the SVD models, with the test RMSE written at the end of each bar, and saves the figure as `hero.png`. The numbers are copied from the outputs of notebooks 02 and 03.

## What we found

The best model (iterative SVD, k = 5) is off by 0.921 stars on hidden ratings, versus 0.954 for the best baseline (user + movie effects): 3.6% better.
